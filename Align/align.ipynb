```python
import numpy as np
import biotite 
import biotite.structure as struc 
from biotite.structure.io import load_structure

def extract_chain_sequences(atom_array):
    residue_starts = struc.get_residue_starts(atom_array)
    unique_res_names = atom_array.res_name[residue_starts]
    return unique_res_names

ref_arr = load_structure('7fct.cif')
pred_arr = load_structure('7fct_model_0.cif')
ref_arr_prot = ref_arr[struc.filter_amino_acids(ref_arr)]
pred_arr_prot = pred_arr[struc.filter_amino_acids(pred_arr)]
ref_arr_prot_A = ref_arr_prot[ref_arr_prot.chain_id == 'A']; ref_seq_A = extract_chain_sequences(ref_arr_prot_A)
ref_arr_prot_B = ref_arr_prot[ref_arr_prot.chain_id == 'B']; ref_seq_B = extract_chain_sequences(ref_arr_prot_B)
pred_arr_prot_A = pred_arr_prot[pred_arr_prot.chain_id == 'A1']; pred_seq_A = extract_chain_sequences(pred_arr_prot_A)
pred_arr_prot_B = pred_arr_prot[pred_arr_prot.chain_id == 'B1']; pred_seq_B = extract_chain_sequences(pred_arr_prot_B)

ref_identifiers = set((r.chain_id+'1', r.res_id, r.atom_name) for r in ref_arr_prot)
pred_mask = np.array([(p.chain_id, p.res_id, p.atom_name) in ref_identifiers for p in pred_arr_prot], dtype=bool)
pred_arr_prot_cleaned = pred_arr_prot[pred_mask]

pred_arr_prot_aligned, transformation = struc.superimpose(
     ref_arr_prot, pred_arr_prot_cleaned#, atom_mask=(xxx.atom_name == "CA")
)
struc.rmsd(ref_arr_prot, pred_arr_prot_aligned)
```
